Retain only the review and label columns.
2. Split the data into training and validation, testing

````python
# 1. Clean and preprocess the dataset
# Assuming 'Series_Title' is the column containing movie titles,
# and 'Overview' is the column containing the movie review.
# Also assuming 'IMDB_Rating' indicates positive/negative sentiment
# where >=7.0 is positive and <7.0 is negative

# Create sentiment column based on IMDB_Rating
df['sentiment'] = df['IMDB_Rating'].apply(lambda x: 1 if float(x) >= 7.0 else 0)

# Retain only the review (Overview) and label (sentiment) columns
df = df[['Overview', 'sentiment']]

# Rename 'Overview' to 'review' for consistency
df = df.rename(columns={'Overview': 'review'})

# 2. Split the data into training, validation, and testing sets
from sklearn.model_selection import train_test_split

# Split into training and remaining data (validation + testing)
train_df, remaining_df = train_test_split(df, test_size=0.3, random_state=42)

# Split remaining data into validation and testing sets
validation_df, test_df = train_test_split(remaining_df, test_size=0.5, random_state=42)

# Print the shapes of the datasets to verify the split
print("Train dataset shape:", train_df.shape)
print("Validation dataset shape:", validation_df.shape)
print("Test dataset shape:", test_df.shape)
````

````python
# Step 1: Install and set up Kaggle API
!pip install kaggle
!mkdir -p ~/.kaggle
!cp kaggle.json ~/.kaggle/
!chmod 600 ~/.kaggle/kaggle.json

# Step 2: Download the dataset
!kaggle datasets download -d harshitshankhdhar/imdb-dataset-of-top-1000-movies-and-tv-shows

# Step 3: Extract the dataset
import zipfile

dataset_zip = "imdb-dataset-of-top-1000-movies-and-tv-shows.zip"
with zipfile.ZipFile(dataset_zip, "r") as zip_ref:
    zip_ref.extractall("imdb-dataset")

# Step 4: Load dataset with Pandas
import pandas as pd

file_path = "imdb-dataset/imdb_top_1000.csv"  # Update with actual filename if different
df = pd.read_csv(file_path)

# Step 5: Verify dataset
print(df.head())
print(df.info())

# Step 2: Data Preprocessing
# Assuming 'Series_Title' is the column containing movie titles,
# and 'Overview' is the column containing the movie review.
# Also assuming 'IMDB_Rating' indicates positive/negative sentiment
# where >=7.0 is positive and <7.0 is negative

# Create sentiment column based on IMDB_Rating
df['sentiment'] = df['IMDB_Rating'].apply(lambda x: 1 if float(x) >= 7.0 else 0)

# Retain only the review (Overview) and label (sentiment) columns
df = df[['Overview', 'sentiment']]

# Rename 'Overview' to 'review' for consistency
df = df.rename(columns={'Overview': 'review'})

# 2. Split the data into training, validation, and testing sets
from sklearn.model_selection import train_test_split

# Split into training and remaining data (validation + testing)
train_df, remaining_df = train_test_split(df, test_size=0.3, random_state=42)

# Split remaining data into validation and testing sets
validation_df, test_df = train_test_split(remaining_df, test_size=0.5, random_state=42)

# Print the shapes of the datasets to verify the split
print("Train dataset shape:", train_df.shape)
print("Validation dataset shape:", validation_df.shape)
print("Test dataset shape:", test_df.shape)

# Step 3: Model Selection and Tokenization
# 1. Select a pre-trained model
!pip install transformers -q
from transformers import AutoTokenizer, AutoModelForSequenceClassification

model_name = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2) # 2 labels for sentiment (positive/negative)

# 2. Tokenize the dataset
def tokenize_function(examples):
    # Convert the Pandas Series to a list of strings
    return tokenizer(examples['review'].tolist(), padding="max_length", truncation=True, max_length=256)
    # The .tolist() method converts the Series to a list, which is the expected input type for the tokenizer

train_encodings = tokenize_function(train_df)
val_encodings = tokenize_function(validation_df)
test_encodings = tokenize_function(test_df)

import torch

class IMDbDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx])
        return item

    def __len__(self):
        return len(self.labels)

train_dataset = IMDbDataset(train_encodings, train_df['sentiment'].tolist())
val_dataset = IMDbDataset(val_encodings, validation_df['sentiment'].tolist())
test_dataset = IMDbDataset(test_encodings, test_df['sentiment'].tolist())

# Step 4: Fine-Tune the Model
from transformers import TrainingArguments, Trainer
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

# Define metrics
def compute_metrics(pred):
    labels = pred.label_ids
    preds = pred.predictions.argmax(-1)
    precision, recall, f1, _ = precision_recall_fscore_support(labels, preds, average='weighted')
    acc = accuracy_score(labels, preds)
    return {
        'accuracy': acc,
        'f1': f1,
        'precision': precision,
        'recall': recall
    }

# Define training arguments
training_args = TrainingArguments(
    output_dir='./results',
    num_train_epochs=2,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=64,
    warmup_steps=500,
    weight_decay=0.01,
    logging_dir='./logs',
    logging_steps=10,
    evaluation_strategy="epoch",
    learning_rate=5e-5,
    load_best_model_at_end=True,
    save_strategy="epoch"
)

# Create Trainer instance
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    compute_metrics=compute_metrics,
)

# Train the model
trainer.train()
````

#Part 2: API Development and Testing (5 points)
Step 6: Set Up the Backend API (1 point)
1. Use FastAPI or Flask, Express, Nest Nodejs to create an API.
2. Define a POST endpoint (/analyze/) that:

  - Accepts:
    - text: The input text for sentiment analysis.
    - model: A parameter specifying the model to use (custom or llama).
  - Returns:
    - Sentiment (positive or negative).
    - Confidence score.

````python
!pip install fastapi uvicorn pyngrok
````

````python
import os
import uvicorn
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from transformers import pipeline
import requests
import nest_asyncio
from pyngrok import ngrok
from google.colab import userdata

from fastapi import FastAPI
from fastapi.middleware.cors import CORSMiddleware

app = FastAPI()

app.add_middleware(
     CORSMiddleware,
     allow_origins=["*"],  # Allows all origins - replace with your React app's origin for production
     allow_credentials=True,
     allow_methods=["*"],  # Allows all methods
     allow_headers=["*"],  # Allows all headers
 )

os.environ ["CROQ_API_KEY"] = userdata.get('croqAPIkey')


# Load your fine-tuned Hugging Face model
classifier = pipeline("sentiment-analysis", model="erkkatt/imdb-sentiment-classifier")

# Function to interact with the Groq Cloud API for the Llama 3 model
def llama3_inference(input_text: str):
    # Correct Groq API endpoint (replace with the correct one)
    url = "https://api.groq.com/openai/v1/chat/completions"
    headers = {
        "Authorization": "Bearer " + os.environ["CROQ_API_KEY"],
        "Content-Type": "application/json"
    }
    data = {
        "model": "llama-3.3-70b-versatile",
      "messages": [{
          "role": "user",
          "content": input_text
      }]
    }

    response = requests.post(url, headers=headers, json=data)

    if response.status_code == 200:
        return response.json()  # Return the response as a JSON
    else:
        raise HTTPException(status_code=500, detail=f"Error with Groq API: {response.text}")


class InputData(BaseModel):
    text: str
    model: str = "custom"  # Default model is "custom"

@app.post("/analyze/")
async def analyze_sentiment(input_data: InputData):
    try:
        if input_data.model == "custom":
            # Use your Hugging Face fine-tuned model
            results = classifier(input_data.text)
            sentiment = results[0]["label"]
            confidence = results[0]["score"]
            return {
                "sentiment": sentiment,
                "confidence": confidence
            }
        elif input_data.model == "llama3":
            # Use the Llama 3 model from Groq API
            result = llama3_inference(input_data.text)
            return result
        else:
            raise HTTPException(status_code=400, detail="Invalid model specified")
    except Exception as e:
        raise HTTPException(status_code=500, detail=f"Error during analysis: {str(e)}")


# Run the FastAPI app
def run():
    # Start the ngrok tunnel
    public_url = ngrok.connect(8000)
    print(f"FastAPI is available at: {public_url}")

    nest_asyncio.apply()  # Apply nest_asyncio patch
    config = uvicorn.Config(app, host="0.0.0.0", port=8000)
    server = uvicorn.Server(config)
    server.run()  # Start the server using server.run()

# Run the FastAPI server
run()
````
